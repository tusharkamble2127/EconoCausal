```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../data/raw/retail_price.csv")

# Display the first five rows
print("First 5 rows:")
display(df.head())

# Display the number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# STEP 2: DATA QUALITY CHECKS

# 1. Check the data types of all columns
print("========== DATA TYPES ==========")
df.info()

# 2. Check missing values
print("\n========== MISSING VALUES ==========")
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])

if missing_values.sum() == 0:
    print("No missing values found.")

# 3. Check duplicate rows
print("\n========== DUPLICATE RECORDS ==========")
print("Number of duplicate rows:", df.duplicated().sum())

# 4. Check numerical statistics
print("\n========== NUMERICAL SUMMARY ==========")
display(df.describe().T)

# 5. Display all column names clearly
print("\n========== ALL COLUMNS ==========")

for column in df.columns:
    print(column)
    
```

```python
# STEP 3: BASIC EXPLORATORY DATA ANALYSIS

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Verify missing values across all columns
print("Missing values in each column:")
print(df.isnull().sum().sort_values(ascending=False))

# 2. Display all column names
print("\nAll dataset columns:")
print(df.columns.tolist())

# 3. Summary statistics for important variables
important_columns = [
    "qty",
    "total_price",
    "unit_price",
    "freight_price",
    "customers"
]

print("\nSummary statistics:")
display(df[important_columns].describe().T)

# 4. Distribution of unit prices
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="unit_price", bins=30, kde=True)
plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 5. Distribution of quantity
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="qty", bins=30, kde=True)
plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 6. Relationship between unit price and quantity
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="unit_price", y="qty")
plt.title("Unit Price vs Quantity")
plt.xlabel("Unit Price")
plt.ylabel("Quantity")
plt.tight_layout()
plt.show()
```

```python
# STEP 4: TIME-BASED EDA

import pandas as pd
import matplotlib.pyplot as plt

# Convert month_year to datetime
df["month_year"] = pd.to_datetime(
    df["month_year"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Check whether date conversion succeeded
print("Invalid dates:", df["month_year"].isna().sum())

# Sort records by date
df = df.sort_values("month_year")

# Calculate monthly average price and quantity
monthly_data = (
    df.groupby("month_year")
      .agg(
          average_unit_price=("unit_price", "mean"),
          average_quantity=("qty", "mean"),
          total_quantity=("qty", "sum"),
          total_revenue=("total_price", "sum")
      )
      .reset_index()
)

print("\nMonthly Summary:")
display(monthly_data.head(10))

# Plot average unit price over time
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_data["month_year"],
    monthly_data["average_unit_price"],
    marker="o"
)
plt.title("Average Unit Price Over Time")
plt.xlabel("Month")
plt.ylabel("Average Unit Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Plot total quantity over time
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_data["month_year"],
    monthly_data["total_quantity"],
    marker="o"
)
plt.title("Total Quantity Over Time")
plt.xlabel("Month")
plt.ylabel("Total Quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# STEP 5: CORRELATION ANALYSIS

import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# Calculate the correlation matrix
correlation_matrix = numeric_df.corr()

# Display correlations with unit_price
print("Correlation with Unit Price:")
display(
    correlation_matrix["unit_price"]
    .sort_values(ascending=False)
    .to_frame(name="Correlation")
)

# Display correlations with quantity
print("\nCorrelation with Quantity:")
display(
    correlation_matrix["qty"]
    .sort_values(ascending=False)
    .to_frame(name="Correlation")
)

# Plot the correlation heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# STEP 6: INVESTIGATE PRICE AND QUANTITY RELATIONSHIPS

import pandas as pd
import numpy as np

# 1. Check whether total price matches unit price multiplied by quantity
df["calculated_total"] = df["unit_price"] * df["qty"]

df["total_price_difference"] = (
    df["total_price"] - df["calculated_total"]
).abs()

print("Price Calculation Check")
print(
    "Mean absolute difference:",
    df["total_price_difference"].mean()
)
print(
    "Maximum absolute difference:",
    df["total_price_difference"].max()
)

# 2. Show example records
print("\nExample price records:")
display(
    df[
        [
            "product_id",
            "month_year",
            "qty",
            "unit_price",
            "total_price",
            "calculated_total",
        ]
    ].head(10)
)

# 3. Count distinct products and months
print("\nNumber of distinct products:", df["product_id"].nunique())
print("Number of distinct months:", df["month_year"].nunique())

# 4. Check unit price variation within each product
product_price_summary = (
    df.groupby("product_id")
      .agg(
          number_of_records=("unit_price", "size"),
          distinct_prices=("unit_price", "nunique"),
          minimum_price=("unit_price", "min"),
          maximum_price=("unit_price", "max"),
      )
      .reset_index()
)

print("\nProduct-level price variation:")
display(product_price_summary.head(10))

print(
    "\nProducts with only one distinct price:",
    (product_price_summary["distinct_prices"] == 1).sum()
)

print(
    "Products with multiple distinct prices:",
    (product_price_summary["distinct_prices"] > 1).sum()
)
```

```python
# STEP 7: INVESTIGATE PRICE CALCULATION DIFFERENCES

# Sort records by the absolute difference
unusual_prices = df.sort_values(
    "total_price_difference",
    ascending=False
)

# Display the 10 records with the largest differences
columns_to_check = [
    "product_id",
    "month_year",
    "qty",
    "unit_price",
    "total_price",
    "calculated_total",
    "total_price_difference"
]

print("10 records with the largest price calculation differences:")

display(unusual_prices[columns_to_check].head(10))
```

```python
# STEP 8: PRODUCT-LEVEL PRICE AND QUANTITY CHANGES

# Summarize price and quantity for each product
product_summary = (
    df.groupby("product_id")
      .agg(
          records=("month_year", "size"),
          distinct_prices=("unit_price", "nunique"),
          average_price=("unit_price", "mean"),
          average_quantity=("qty", "mean"),
          minimum_price=("unit_price", "min"),
          maximum_price=("unit_price", "max"),
      )
      .reset_index()
)

# Calculate the price range for each product
product_summary["price_range"] = (
    product_summary["maximum_price"]
    - product_summary["minimum_price"]
)

print("Product-Level Summary:")
display(product_summary.head(10))

# Summarize price variation across all products
print("\nTotal number of products:", df["product_id"].nunique())

print(
    "Products with changing prices:",
    (product_summary["distinct_prices"] > 1).sum()
)

print(
    "Products with constant prices:",
    (product_summary["distinct_prices"] == 1).sum()
)

print(
    "Average price range across products:",
    product_summary["price_range"].mean()
)
```

```python
# STEP 9: WITHIN-PRODUCT PRICE AND QUANTITY ANALYSIS

import matplotlib.pyplot as plt
import seaborn as sns

# Calculate correlations separately for each product.
# Products with no price variation cannot provide a price-quantity
# correlation, so exclude them from this calculation.

product_correlations = []

for product_id, group in df.groupby("product_id"):

    if group["unit_price"].nunique() > 1 and group["qty"].nunique() > 1:

        correlation = group["unit_price"].corr(group["qty"])

        product_correlations.append({
            "product_id": product_id,
            "records": len(group),
            "price_quantity_correlation": correlation
        })

product_corr_df = pd.DataFrame(product_correlations)

print("Number of products analyzed:", len(product_corr_df))

print("\nCorrelation summary across products:")
display(product_corr_df["price_quantity_correlation"].describe())

print("\nProduct-level correlations:")
display(
    product_corr_df.sort_values(
        "price_quantity_correlation"
    ).head(10)
)

# Visualize the correlations across products
plt.figure(figsize=(9, 5))

sns.histplot(
    data=product_corr_df,
    x="price_quantity_correlation",
    bins=15,
    kde=True
)

plt.axvline(0, linestyle="--")
plt.title("Within-Product Price-Quantity Correlations")
plt.xlabel("Correlation between Unit Price and Quantity")
plt.ylabel("Number of Products")
plt.tight_layout()
plt.show()
```

```python
# STEP 10: SUMMARIZE PRODUCT-LEVEL CORRELATIONS

# Overall summary of product-level correlations
print("Correlation Summary:")
display(
    product_corr_df["price_quantity_correlation"].describe()
)

# Count positive, negative, and near-zero correlations
positive = (
    product_corr_df["price_quantity_correlation"] > 0.05
).sum()

negative = (
    product_corr_df["price_quantity_correlation"] < -0.05
).sum()

near_zero = (
    product_corr_df["price_quantity_correlation"].between(-0.05, 0.05)
).sum()

print("\nPositive correlations:", positive)
print("Negative correlations:", negative)
print("Near-zero correlations:", near_zero)

# Display the 10 most positive correlations
print("\n10 Most Positive Product-Level Correlations:")

display(
    product_corr_df.sort_values(
        "price_quantity_correlation",
        ascending=False
    ).head(10)
)
```

```python
# STEP 13: INSPECT ALL DATASET VARIABLES

# Display every column with its data type
print("========== ALL DATASET COLUMNS ==========")

column_info = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Unique Values": [
        df[col].nunique() for col in df.columns
    ],
    "Missing Values": [
        df[col].isnull().sum() for col in df.columns
    ]
})

display(column_info)

# Display example values for every column
print("\n========== SAMPLE VALUES ==========")

for col in df.columns:
    print(f"\n{col}:")
    print(df[col].dropna().unique()[:5])
```

```python
# STEP 15: FINAL DATASET STRUCTURE CHECK

# 1. Original dataset columns
original_columns = [
    "product_id",
    "product_category_name",
    "month_year",
    "qty",
    "total_price",
    "freight_price",
    "unit_price",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_score",
    "customers",
    "weekday",
    "weekend",
    "holiday",
    "month",
    "year",
    "s",
    "volume",
    "comp_1",
    "ps1",
    "fp1",
    "comp_2",
    "ps2",
    "fp2",
    "comp_3",
    "ps3",
    "fp3",
    "lag_price"
]

print("Original dataset columns:", len(original_columns))
print("Current dataset columns:", len(df.columns))

# 2. Verify that the original columns are present
missing_columns = [
    col for col in original_columns
    if col not in df.columns
]

print("Missing original columns:", missing_columns)

# 3. Verify the price calculation
df["calculated_total"] = df["unit_price"] * df["qty"]

df["total_price_difference"] = (
    df["total_price"] - df["calculated_total"]
).abs()

print(
    "\nRecords with a difference greater than 0.01:",
    (df["total_price_difference"] > 0.01).sum()
)

# 4. Display the potential treatment and outcome
display(
    df[["product_id", "month_year", "unit_price", "qty"]]
    .head(10)
)
```

```python
# STEP 16: CHECK DATA FOR CAUSAL ANALYSIS

# Select the main variables for our project
analysis_columns = [
    "product_id",
    "product_category_name",
    "month_year",
    "unit_price",
    "qty",
    "lag_price",
    "comp_1",
    "comp_2",
    "comp_3",
    "customers"
]

# Check that all required columns exist
missing_analysis_columns = [
    col for col in analysis_columns
    if col not in df.columns
]

print("Missing analysis columns:", missing_analysis_columns)

# Create a separate dataset for initial causal analysis
analysis_df = df[analysis_columns].copy()

# Display dataset information
print("\nAnalysis dataset shape:", analysis_df.shape)

print("\nMissing values:")
display(analysis_df.isnull().sum())

print("\nSummary of price and quantity:")
display(analysis_df[["unit_price", "qty", "lag_price"]].describe())

print("\nNumber of unique products:", analysis_df["product_id"].nunique())

print(
    "Number of unique months:",
    analysis_df["month_year"].nunique()
)

print(
    "\nRecords with non-positive prices:",
    (analysis_df["unit_price"] <= 0).sum()
)

print(
    "Records with non-positive quantities:",
    (analysis_df["qty"] <= 0).sum()
)
```

```python
# STEP 17: INVESTIGATE PRICE VARIATION AND CONTROL VARIABLES

import pandas as pd

# 1. Check price variation for each product
product_price_summary = (
    analysis_df.groupby("product_id")
    .agg(
        number_of_records=("unit_price", "size"),
        unique_prices=("unit_price", "nunique"),
        minimum_price=("unit_price", "min"),
        maximum_price=("unit_price", "max"),
        average_price=("unit_price", "mean"),
        average_quantity=("qty", "mean")
    )
)

product_price_summary["price_range"] = (
    product_price_summary["maximum_price"]
    - product_price_summary["minimum_price"]
)

print("PRODUCT-LEVEL PRICE VARIATION")
print(
    "Products with changing prices:",
    (product_price_summary["unique_prices"] > 1).sum()
)
print(
    "Products with constant prices:",
    (product_price_summary["unique_prices"] == 1).sum()
)

display(product_price_summary.head(10))


# 2. Check how price and quantity move together
print("\nPRICE AND QUANTITY CORRELATION")

correlation = analysis_df[[
    "unit_price",
    "qty",
    "lag_price",
    "comp_1",
    "comp_2",
    "comp_3",
    "customers"
]].corr()

display(correlation.round(3))


# 3. Check how many products appear in each month
print("\nRECORDS PER MONTH")

monthly_records = (
    analysis_df.groupby("month_year")
    .agg(
        number_of_records=("product_id", "size"),
        number_of_products=("product_id", "nunique"),
        average_price=("unit_price", "mean"),
        total_quantity=("qty", "sum")
    )
)

display(monthly_records)
```

```python
# STEP 18: CHECK PRODUCT-MONTH STRUCTURE

# Work with the original analysis variables
panel_df = analysis_df.copy()

# 1. Check duplicate product-month combinations
duplicate_product_months = panel_df.duplicated(
    subset=["product_id", "month_year"]
).sum()

print("Duplicate product-month records:", duplicate_product_months)

# 2. Check how many months each product appears in
product_month_counts = (
    panel_df.groupby("product_id")["month_year"]
    .nunique()
    .sort_values()
)

print("\nPRODUCT COVERAGE")
print("Minimum months per product:", product_month_counts.min())
print("Maximum months per product:", product_month_counts.max())
print("Average months per product:", round(product_month_counts.mean(), 2))

display(product_month_counts.head(10))

# 3. Check whether lag_price matches the previous observed price
panel_df = panel_df.sort_values(
    ["product_id", "month_year"]
).copy()

panel_df["previous_observed_price"] = (
    panel_df.groupby("product_id")["unit_price"].shift(1)
)

panel_df["lag_price_difference"] = (
    panel_df["lag_price"] - panel_df["previous_observed_price"]
).abs()

valid_lag_rows = panel_df["previous_observed_price"].notna()

matching_lag_rows = (
    panel_df.loc[valid_lag_rows, "lag_price_difference"] < 0.01
).sum()

number_of_comparisons = valid_lag_rows.sum()

print("\nLAG PRICE CHECK")
print("Rows compared:", number_of_comparisons)
print("Rows where lag_price matches the previous observed price:",
      matching_lag_rows)

if number_of_comparisons > 0:
    print(
        "Match percentage:",
        round(100 * matching_lag_rows / number_of_comparisons, 2),
        "%"
    )

# 4. Check whether consecutive observations are exactly one month apart
panel_df["months_since_previous"] = (
    panel_df.groupby("product_id")["month_year"]
    .diff()
    .dt.days / 30.4375
)

print("\nTIME GAP CHECK")
display(
    panel_df["months_since_previous"]
    .describe()
    .round(2)
)
```

```python
# STEP 19: INVESTIGATE LAG-PRICE MISMATCHES

# Keep rows where a previous observation exists
lag_check = panel_df.loc[
    panel_df["previous_observed_price"].notna()
].copy()

# Identify mismatches
lag_mismatches = lag_check.loc[
    lag_check["lag_price_difference"] >= 0.01,
    [
        "product_id",
        "month_year",
        "unit_price",
        "lag_price",
        "previous_observed_price",
        "lag_price_difference",
        "months_since_previous"
    ]
].copy()

print("Total lag-price mismatches:", len(lag_mismatches))

# Display the first 15 mismatches
display(lag_mismatches.head(15))

# Summarize mismatch size
if not lag_mismatches.empty:
    print("\nMISMATCH SIZE SUMMARY")
    display(lag_mismatches["lag_price_difference"].describe())

# Check whether mismatches occur after longer time gaps
if not lag_mismatches.empty:
    print("\nTIME GAPS FOR MISMATCHED RECORDS")
    display(lag_mismatches["months_since_previous"].describe())
```

```python
import pandas as pd

df = pd.read_csv(r"D:\projects\EconoCausal-Dynamic-Pricing-DML\data\raw\retail_price.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("Column names:")
print(df.columns.tolist())

display(df.head())
```

```python
# STEP 21: PREPARE INITIAL CAUSAL-ANALYSIS DATASET

import pandas as pd

# Select variables for the initial analysis
causal_columns = [
    "product_id",
    "product_category_name",
    "month_year",
    "unit_price",
    "qty",
    "freight_price",
    "product_score",
    "product_weight_g",
    "product_photos_qty",
    "comp_1",
    "comp_2",
    "comp_3",
    "ps1",
    "ps2",
    "ps3",
    "fp1",
    "fp2",
    "fp3"
]

# Create a separate copy
causal_df = df[causal_columns].copy()

# Check the dataset
print("Causal-analysis dataset shape:", causal_df.shape)

print("\nMissing values:")
print(causal_df.isnull().sum().sum())

print("\nUnique products:", causal_df["product_id"].nunique())

print("\nUnique months:", causal_df["month_year"].nunique())

print("\nTreatment variable: unit_price")
print("Outcome variable: qty")

print("\nFirst five records:")
display(causal_df.head())

print("\nTreatment and outcome summary:")
display(causal_df[["unit_price", "qty"]].describe())
```

```python
# STEP 22: INVESTIGATE COMPETITOR-PRICE VARIABLES

competitor_columns = ["comp_1", "comp_2", "comp_3"]

for col in competitor_columns:
    matches = (causal_df[col] == causal_df["unit_price"]).sum()

    print(f"\n{col}")
    print("Matches with unit_price:", matches)
    print(
        "Percentage of records:",
        round(matches / len(causal_df) * 100, 2),
        "%"
    )
    print("Unique values:", causal_df[col].nunique())

# Check whether competitor prices are identical to one another
print("\nCompetitor-price equality")

print(
    "comp_1 equals comp_2:",
    (causal_df["comp_1"] == causal_df["comp_2"]).sum()
)

print(
    "comp_2 equals comp_3:",
    (causal_df["comp_2"] == causal_df["comp_3"]).sum()
)

print(
    "All three competitor prices are equal:",
    (
        (causal_df["comp_1"] == causal_df["comp_2"])
        & (causal_df["comp_2"] == causal_df["comp_3"])
    ).sum()
)

# Display records where every competitor price equals the product price
same_price_records = causal_df.loc[
    (causal_df["comp_1"] == causal_df["unit_price"])
    & (causal_df["comp_2"] == causal_df["unit_price"])
    & (causal_df["comp_3"] == causal_df["unit_price"]),
    ["product_id", "month_year", "unit_price",
     "comp_1", "comp_2", "comp_3"]
]

print("\nRecords where all competitor prices equal unit_price:")
display(same_price_records.head(10))
```

```python
# STEP 23: CHECK WITHIN-PRODUCT PRICE VARIATION

import pandas as pd

# Work with a clean copy
price_check_df = causal_df.copy()

# Calculate within-product price variation
price_variation = (
    price_check_df.groupby("product_id")
    .agg(
        observations=("unit_price", "size"),
        unique_prices=("unit_price", "nunique"),
        minimum_price=("unit_price", "min"),
        maximum_price=("unit_price", "max"),
        average_price=("unit_price", "mean"),
        price_std=("unit_price", "std")
    )
)

price_variation["price_range"] = (
    price_variation["maximum_price"]
    - price_variation["minimum_price"]
)

print("PRODUCT-LEVEL PRICE VARIATION")

print(
    "Products with changing prices:",
    (price_variation["unique_prices"] > 1).sum()
)

print(
    "Products with constant prices:",
    (price_variation["unique_prices"] == 1).sum()
)

print("\nWithin-product price variation summary:")
display(price_variation.describe().round(3))

print("\nProducts with the smallest price ranges:")
display(price_variation.sort_values("price_range").head(10))

print("\nProducts with the largest price ranges:")
display(price_variation.sort_values("price_range", ascending=False).head(10))


# Check overall and within-product price variation
overall_price_std = price_check_df["unit_price"].std()

within_product_price_std = (
    price_check_df.groupby("product_id")["unit_price"]
    .std()
    .mean()
)

print("\nOVERALL PRICE STANDARD DEVIATION:", round(overall_price_std, 3))
print(
    "Average within-product price standard deviation:",
    round(within_product_price_std, 3)
)
```

```python
# STEP 60: DEFINE THE DOWHY CAUSAL MODEL

from dowhy import CausalModel

# Build the graph using observed variable names
graph = """
digraph {
    comp_1 -> unit_price;
    comp_2 -> unit_price;
    comp_3 -> unit_price;

    comp_1 -> qty;
    comp_2 -> qty;
    comp_3 -> qty;

    freight_price -> unit_price;
    freight_price -> qty;

    product_score -> unit_price;
    product_score -> qty;

    product_weight_g -> unit_price;
    product_weight_g -> qty;

    product_photos_qty -> unit_price;
    product_photos_qty -> qty;

    product_id -> unit_price;
    product_id -> qty;

    month_year -> unit_price;
    month_year -> qty;

    unit_price -> qty;
}
"""

# Use a copy with categorical identifiers represented as strings
dowhy_df = model_df.copy()

dowhy_df["product_id"] = dowhy_df["product_id"].astype(str)
dowhy_df["month_year"] = dowhy_df["month_year"].astype(str)

# Define the causal model
causal_model = CausalModel(
    data=dowhy_df,
    treatment="unit_price",
    outcome="qty",
    graph=graph,
)

print("DoWhy causal model created successfully!")
print("Treatment:", causal_model._treatment)
print("Outcome:", causal_model._outcome)
print("Rows in model data:", len(dowhy_df))
```

```python
# STEP 62: ESTIMATE AN INITIAL PRICE EFFECT
# Using product and month fixed effects

import statsmodels.formula.api as smf

# Select observed variables for the baseline model
baseline_df = analysis_df.copy()

# Build the regression formula
formula = """
qty ~ unit_price
    + freight_price
    + comp_1
    + comp_2
    + comp_3
    + product_score
    + product_weight_g
    + product_photos_qty
    + C(product_id)
    + C(month_year)
"""

# Fit the baseline model
baseline_model = smf.ols(
    formula=formula,
    data=baseline_df
).fit(
    cov_type="cluster",
    cov_kwds={"groups": baseline_df["product_id"]}
)

print("INITIAL PRICE EFFECT ESTIMATE\n")

print("Price coefficient:", baseline_model.params["unit_price"])
print("Clustered standard error:", baseline_model.bse["unit_price"])
print("P-value:", baseline_model.pvalues["unit_price"])
print("95% confidence interval:")
print(baseline_model.conf_int().loc["unit_price"])

print("\nNumber of observations:", int(baseline_model.nobs))
print("R-squared:", baseline_model.rsquared)
```

```python
# STEP 63: DIAGNOSE THE DESIGN MATRIX

import numpy as np

# Extract the fitted regression design matrix
X = baseline_model.model.exog

# Calculate matrix dimensions and rank
matrix_rank = np.linalg.matrix_rank(X)
number_of_columns = X.shape[1]

print("DESIGN MATRIX DIAGNOSTICS\n")

print("Number of observations:", X.shape[0])
print("Number of design columns:", number_of_columns)
print("Matrix rank:", matrix_rank)
print("Rank deficiency:", number_of_columns - matrix_rank)

print("\nIs the design matrix full rank?")
print(matrix_rank == number_of_columns)

# Check for columns with no variation
design_df = baseline_model.model.data.orig_exog

constant_columns = [
    col for col in design_df.columns
    if design_df[col].nunique() <= 1
]

print("\nConstant design columns:")
print(constant_columns)
```

```python
# STEP 64: IDENTIFY LINEAR DEPENDENCIES

import numpy as np
import pandas as pd
from scipy.linalg import qr

# Get the regression design matrix and its column names
X = baseline_model.model.exog
column_names = baseline_model.model.exog_names

# Perform QR decomposition with column pivoting
Q, R, pivot = qr(X, mode="economic", pivoting=True)

# Determine numerical rank
rank = np.linalg.matrix_rank(X)

# Columns selected as independent
independent_columns = [
    column_names[i] for i in pivot[:rank]
]

# Remaining columns are redundant relative to the selected basis
redundant_columns = [
    column_names[i] for i in pivot[rank:]
]

print("LINEAR DEPENDENCY ANALYSIS\n")

print("Matrix rank:", rank)
print("Total columns:", len(column_names))

print("\nIndependent columns:", len(independent_columns))
print("Redundant columns:", len(redundant_columns))

print("\nRedundant columns identified by QR pivoting:")
for column in redundant_columns:
    print("-", column)
```

```python
# STEP 69: SET UP CATEGORICAL AND NUMERICAL PREPROCESSING

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Numerical and categorical control columns
numeric_controls = [
    "freight_price",
    "comp_1",
    "comp_2",
    "comp_3",
]

categorical_controls = [
    "product_id",
    "month_year",
]

# Preprocessing for numerical variables
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# Preprocessing for categorical variables
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)

# Combine both preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_controls),
        ("categorical", categorical_transformer, categorical_controls),
    ]
)

print("PREPROCESSING CONFIGURATION")
print("-" * 40)

print("Numerical controls:", numeric_controls)
print("Categorical controls:", categorical_controls)
print("Numerical scaling: StandardScaler")
print("Categorical encoding: OneHotEncoder")
print("Unknown categories: ignored")
print("\nPreprocessor configured successfully!")
```

```python
# STEP 70B: CONFIGURE DML MACHINE-LEARNING MODELS

from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.base import clone

# Outcome model: predicts quantity sold
ml_l = Pipeline(
    steps=[
        ("preprocessor", clone(preprocessor)),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=6,
                min_samples_leaf=5,
                random_state=42,
                n_jobs=-1,
            ),
        ),
    ]
)

# Treatment model: predicts unit price
ml_m = Pipeline(
    steps=[
        ("preprocessor", clone(preprocessor)),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=6,
                min_samples_leaf=5,
                random_state=42,
                n_jobs=-1,
            ),
        ),
    ]
)

print("DML LEARNERS CONFIGURED")
print("-" * 40)

print("Outcome learner:", type(ml_l.named_steps["model"]).__name__)
print("Treatment learner:", type(ml_m.named_steps["model"]).__name__)

print(
    "Outcome trees:",
    ml_l.named_steps["model"].n_estimators,
)
print(
    "Treatment trees:",
    ml_m.named_steps["model"].n_estimators,
)

print("\nBoth learners configured successfully!")
```

```python
# STEP 98: PREPARE THE PANEL DATASET FOR DOUBLEML

import pandas as pd
from doubleml import DoubleMLPanelData

# Select the outcome, treatment, panel identifiers and time-varying controls.
panel_controls = [
    "freight_price",
    "comp_1",
    "comp_2",
    "comp_3",
    "ps1",
    "ps2",
    "ps3",
    "fp1",
    "fp2",
    "fp3"
]

panel_model_df = analysis_df[
    [
        "product_id",
        "month_year",
        "qty",
        "unit_price"
    ] + panel_controls
].copy()

# Sort observations by product and time.
panel_model_df = (
    panel_model_df
    .sort_values(["product_id", "month_year"])
    .reset_index(drop=True)
)

# Create numerical month indicators.
month_dummies = pd.get_dummies(
    panel_model_df["month_year"].dt.strftime("%Y-%m"),
    prefix="period",
    drop_first=True,
    dtype=int
)

panel_model_df = pd.concat(
    [panel_model_df, month_dummies],
    axis=1
)

# Define the complete set of numerical control variables.
panel_x_cols = panel_controls + month_dummies.columns.tolist()

# Create the DoubleML panel data object.
panel_data_obj = DoubleMLPanelData(
    data=panel_model_df,
    y_col="qty",
    d_cols="unit_price",
    t_col="month_year",
    id_col="product_id",
    x_cols=panel_x_cols,
    static_panel=True,
    datetime_unit="M"
)

# Validate the prepared dataset.
print("PANEL DATASET PREPARATION")
print("=" * 55)

print("\nDATA SHAPE")
print("Rows:", panel_model_df.shape[0])
print("Columns:", panel_model_df.shape[1])

print("\nPANEL STRUCTURE")
print("Unique products:", panel_data_obj.n_ids)
print("Unique months:", panel_data_obj.n_t_periods)
print("Observations:", panel_data_obj.n_obs)
print("Static panel:", panel_data_obj.static_panel)

print("\nMODEL VARIABLES")
print("Outcome:", panel_data_obj.y_col)
print("Treatment:", panel_data_obj.d_cols)
print("Number of controls:", len(panel_data_obj.x_cols))
print("Monthly indicators:", len(month_dummies.columns))

print("\nDATA VALIDATION")
print(
    "Missing values in outcome, treatment and controls:",
    panel_model_df[
        ["qty", "unit_price"] + panel_x_cols
    ].isna().sum().sum()
)

print(
    "Duplicate product-month rows:",
    panel_model_df.duplicated(
        subset=["product_id", "month_year"]
    ).sum()
)

print(
    "All controls numeric:",
    all(
        pd.api.types.is_numeric_dtype(panel_model_df[col])
        for col in panel_x_cols
    )
)
```

```python
# STEP 99: CREATE THE PANEL DOUBLE MACHINE LEARNING ESTIMATOR

from doubleml import DoubleMLPLPR
from sklearn.ensemble import RandomForestRegressor

# Outcome learner: predicts quantity sold.
panel_ml_l = RandomForestRegressor(
    n_estimators=100,
    max_depth=6,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

# Treatment learner: predicts unit price.
panel_ml_m = RandomForestRegressor(
    n_estimators=100,
    max_depth=6,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

# Create a panel-specific DML estimator.
dml_panel_cre = DoubleMLPLPR(
    obj_dml_data=panel_data_obj,
    ml_l=panel_ml_l,
    ml_m=panel_ml_m,
    n_folds=5,
    n_rep=1,
    score="partialling out",
    approach="cre_general"
)

# Inspect the estimator without fitting it yet.
print("PANEL DOUBLEML ESTIMATOR")
print("=" * 55)

print("Estimator:", type(dml_panel_cre).__name__)
print("Approach: cre_general")
print("Score: partialling out")
print("Cross-fitting folds:", dml_panel_cre.n_folds)
print("Observations:", dml_panel_cre.n_obs)

print("\nOUTCOME LEARNER")
print(type(panel_ml_l).__name__)
print("Trees:", panel_ml_l.n_estimators)
print("Maximum depth:", panel_ml_l.max_depth)

print("\nTREATMENT LEARNER")
print(type(panel_ml_m).__name__)
print("Trees:", panel_ml_m.n_estimators)
print("Maximum depth:", panel_ml_m.max_depth)

print("\nTRANSFORMED DATA")
print(
    "Shape:",
    dml_panel_cre.data_transform.data.shape
)
print(
    "Transformed control variables:",
    len(dml_panel_cre.data_transform.x_cols)
)
```

```python
# STEP 100: FIT THE PANEL DOUBLE MACHINE LEARNING MODEL

print("FITTING PANEL DOUBLEML MODEL")
print("=" * 55)

dml_panel_cre.fit()

print("\nPANEL DML ESTIMATION RESULTS")
print("=" * 55)

print(dml_panel_cre.summary)

print("\nMODEL DETAILS")
print("-" * 55)
print("Estimator:", type(dml_panel_cre).__name__)
print("Approach:", dml_panel_cre.approach)
print("Observations:", dml_panel_cre.n_obs)
print("Cross-fitting folds:", dml_panel_cre.n_folds)

print("\nESTIMATED PRICE EFFECT")
print("-" * 55)

print("Coefficient:", round(float(dml_panel_cre.coef[0]), 6))
print("Standard error:", round(float(dml_panel_cre.se[0]), 6))
print("P-value:", round(float(dml_panel_cre.pval[0]), 6))

panel_ci = dml_panel_cre.confint()

print(
    "95% confidence interval:",
    (
        round(float(panel_ci.iloc[0, 0]), 6),
        round(float(panel_ci.iloc[0, 1]), 6)
    )
)
```

## Causal Graph and DoWhy Identification

### 1. Objective

The objective is to investigate the effect of `unit_price` (treatment) on `qty` (outcome) using causal analysis.

### 2. Variables and Assumed Relationships

| Variable | Description |
|---|---|
| `unit_price` | Product selling price; treatment |
| `qty` | Quantity sold; outcome |
| `comp_1`, `comp_2`, `comp_3` | Competitor prices |
| `freight_price` | Product freight price |
| `product_score` | Product score |
| `product_weight_g` | Product weight |
| `product_photos_qty` | Number of product photos |
| `product_id` | Product identifier |
| `month_year` | Observation month |

The graph assumes that competitor prices and freight prices may influence both our product's price and its quantity sold. Product characteristics, product identity, and time may also help explain differences in pricing and sales.

The principal relationship being investigated is `unit_price -> qty`.

These relationships are assumptions used to construct the causal model, not facts established by correlation alone.

### 3. DoWhy Identification

DoWhy was used to construct the causal model and identify an estimand from the specified graph.

The previously reported estimand was `NONPARAMETRIC_ATE`, identified using the backdoor criterion.

The reported adjustment set included `comp_1`, `comp_2`, `comp_3`, `freight_price`, `product_score`, `product_weight_g`, `product_photos_qty`, `product_id`, and `month_year`.

The backdoor approach attempts to account for observed variables that could create a misleading relationship between price and quantity.

### 4. Limitations

The dataset is observational. Causal conclusions depend on the correctness of the graph, appropriate control-variable selection, and assumptions about unmeasured confounding.

Product characteristics such as weight and photo count do not vary within products in this dataset, limiting estimation of their separate effects when product fixed effects are included.

DoWhy's identification result does not, by itself, prove that the true causal effect has been identified.

### 5. Conclusion

The causal graph represents the hypothesized relationships between retail price and quantity sold. DoWhy identifies a backdoor adjustment strategy under the specified assumptions, providing a foundation for our regression and Double Machine Learning analyses.
